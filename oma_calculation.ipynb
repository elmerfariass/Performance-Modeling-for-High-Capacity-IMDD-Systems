## 1. Calculation of the Outer OMA ($OMA_{outer}$)

The `calculate_oma_outer` function is responsible for computing the **Outer Optical Modulation Amplitude** ($OMA_{outer}$), which is a parameter that quantifies the optical signal power variation in intensity-modulation direct-detection (IMDD) systems.

### Expression

The function implements **Equation (7)** from the reference paper, given by:

$$OMA_{outer} = 2 \cdot P_{tx,avg} \cdot \frac{ER - 1}{ER + 1}$$

Where:
* $P_{tx,avg}$ is the **Average Transmitted Optical Power** (typically provided in mW).
* $ER$ is the **Extinction Ratio**, expressed in **linear scale**.

Since the Extinction Ratio is commonly given in decibels ($ER_{dB}$), the function internally performs the conversion to the linear scale before applying the formula:
> $$ER = 10^{\frac{ER_{dB}}{10}}$$


```python
import numpy as np

def calculate_oma_outer(p_tx_avg_mw, er_db):
    """
    Calculates the Outer OMA (Equation 7).

    Parameters:
    p_tx_avg_mw (float): Average transmitted optical power [mW].
    er_db (float): Extinction Ratio [dB].

    Returns:
    float: Outer OMA [mW].
    """
    
    er_lin = 10**(er_db / 10.0)  # Conversão de dB para escala linear

    # Cálculo da OMA outer usando a equação(7)
    
    oma_outer = 2 * p_tx_avg_mw * (er_lin - 1) / (er_lin + 1)
    return oma_outer
```

```python
import numpy as np

# --- Execução do Teste de Validação ---

print("--- Starting Validation Test for calculate_oma_outer ---")

# 1. Parâmetros de teste (Valores típicos de sistemas IMDD)
p_tx_avg_test = 1.0  # 1.0 mW (0 dBm)
er_db_test = 6.0     # 6.0 dB

# 2. Executa a função
oma_result = calculate_oma_outer(p_tx_avg_test, er_db_test)

# 3. Cálculo analítico esperado para checagem manual:
# ER_linear = 10^(6/10) = 3.981
# OMA = 2 * 1 * (3.981 - 1) / (3.981 + 1) = 2 * 2.981 / 4.981 ≈ 1.197 mW
expected_val = 1.1969

# 4. Exibe os resultados na tela
print(f"Input Average Power: {p_tx_avg_test} mW")
print(f"Input Extinction Ratio: {er_db_test} dB")
print("-" * 40)
print(f"Calculated Outer OMA: {oma_result:.4f} mW")
print(f"Expected Outer OMA:   {expected_val:.4f} mW")

# 5. Verificação automática de tolerância
if np.isclose(oma_result, expected_val, rtol=1e-3):
    print("\n SUCCESS! The OMA calculation matches the theoretical expectation.")
else:
    print("\n ERROR! There is a mismatch. Check the dB to linear conversion.")
```
